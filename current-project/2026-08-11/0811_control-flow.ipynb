```python
#beak: 반복문 조건과 관계없이 반복문을 탈출할 때 사용
num=1
while num<10:
    print(num)
    if num%5==0:
        break
    num=num+1
print("반복문이 종료되었습니다.")
```

```python
#continue: 현재 수행중인 반복의 나머지 명령들을 중단하고 다음 단계의 반복을 진행
num=0
while num<10:
    num= num+1
    if num%3 == 0 or num% 5 ==0: # 3이나 5의 배수는 출력하지 않음
        continue
    print(num) 
print("반복문이 종료되었습니다.")
```

```python
# 양의 정수 n을 입력받은 다음 1부터 n까지의 합을 구하기
n = int(input("양의 정수를 입력하세요: "))
sum = 0

for num in range(1, n+1):
    sum = sum + num

print("1부터 %s까지의 합 = %s" % (n, sum))
```

```python
#실습문제2
A=0
B=0
C=0
D=0
F=0

for s in range(1, 6):
    num = int(input("%s번째 숫자를 입력하세요:" %s))
    if (num>=90):
        A+=1
    elif (num>=80):
        B+=1
    elif (num>=70):
        C+=1
    elif (num>=60):
        D+=1
    else:
        F+=1
print("A:%s명 B:%s명 C:%s명 D:%s명 F:%s명" %(A, B, C, D, F))
```

```python
#실습문제3
plus=0
minus=0
total=0
avg=0

for s in range(1, 6):
    num = int(input("%s번째 숫자: " %s))

    if num>0:
        plus+=1
    elif num<0:
        minus+=1

    total+=num

avg= total/5


print("양수의 개수:", plus)
print("음수의 개수:", minus)
print("전체 합:", total)
print("평균:", avg)
```

```python
#실습문제5
plus=0
minus=0
total=0
num=1

while num!=0:
    num=int(input("숫자를 입력하세요.:"))
    total+=num
    
    if num>0:
        print("양수 입니다.")
        plus+=1
    elif num<0:
        print("음수 입니다.")
        minus+=1
print("입력 종료!","\n양수:",plus,"개","\n음수:",minus,"개","\n전체 합",total)
```

```python
positive=0
negative=0
sum=0

while True:
    num=int(input("숫자를 입력하세요:"))

    if num>0:
        print("양수입니다.")
        positive+=1
    elif num<0:
        print("음수입니다.")
        negative+=1
    sum+=num
    if num==0:
        break

print("입력 종료!")
print("양수:",positive,"개")
print("음수:",negative,"개")
print("전체 합:",sum)
```

```python
#실습문제6
total=0
price=1
dis=0

while price!=0:
    price=int(input("상품가격을 입력하세요:"))
    total+=price
    
    if total>=50000:
        dis=20
    elif total>=30000:
        dis=10
    elif total>=10000:
        dis=5
    else:
        dis=0
        
print("할인 전 금액",total,"원", "\n할인율:",dis,"%","\n할인 금액:",total*dis/100,"원")
```

```python
tot=0
while True:
    price=int(input("상품 가격을 입력하세요:"))

    if price==0:
        break
    tot+=price
if tot>=50000:
    discount_rate=0.2
elif tot>=30000:
    discount_rate=0.1
elif tot>=10000:
    discount_rate=0.05
else:
    discount_rate=0

discount=tot*discount_rate

print("=====결제 결과=====")
print("할인 전 금액:",tot,"원")
print("할인율:",discount_rate*100,"%")
print("할인 금액:",discount,"원")
```
